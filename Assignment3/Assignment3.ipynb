Visualize the data using Python by plotting the graphs for assignment no. 1 and 2. Consider a suitable data set. Use Scatter plot, Bar plot, Box plot, Pie chart, Line Chart.

```python
import pandas as pd

data = {
    'Month': ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun'],
    'Sales': [1200, 1500, 1100, 1800, 2000, 2200],
    'Profit': [300, 450, 250, 500, 600, 700],
    'Expenses': [900, 1050, 850, 1300, 1400, 1500],
    'Customer_Count': [80, 95, 70, 110, 130, 150]
}
df=pd.DataFrame(data)
print(df)
```

Scatter Plot

```python
import matplotlib.pyplot as plt

plt.scatter(df['Customer_Count'], df['Sales'])
plt.xlabel('Customer Count')
plt.ylabel('Sales')
plt.title('Impact of Customer Count on Sales Performance')
plt.show()

```

Bar Plot

```python
plt.bar(df['Month'], df['Sales'])
plt.xlabel('Month')
plt.ylabel('Sales')
plt.title('Monthly Sales Distribution')
plt.show()
```

Box Plot


```python
import matplotlib.pyplot as plt

plt.boxplot([data['Sales'], data['Profit'], data['Expenses']],labels=['Sales','Profit','Expenses'])
plt.title('Distribution Analysis of Sales, Profit, and Expenses')
plt.show()
```

Pie Chart

```python
import matplotlib.pyplot as plt

average_values = df[['Sales', 'Profit', 'Expenses']].mean()

plt.pie(average_values, labels=average_values.index, autopct='%1.1f%%')
plt.title('Average Financial Contribution Analysis')
plt.show()
```

Line Chart

```python
import matplotlib.pyplot as plt

plt.plot(df['Month'], df['Sales'], label='Sales')
plt.plot(df['Month'], df['Profit'],label='Profit')
plt.plot(df['Month'], df['Expenses'],label='Expenses')

plt.xlabel('Month')
plt.ylabel('Amount')
plt.title('Financial Performance Trend Over Time')
plt.legend()
plt.show()
```
