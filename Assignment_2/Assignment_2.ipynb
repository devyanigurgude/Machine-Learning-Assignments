Performing Operations Using User Defined Functions

```python
def calc_mean(numbers):
    return sum(numbers) / len(numbers)

def calc_min(numbers):
    return min(numbers)

def calc_max(numbers):
    return max(numbers)

def calc_median(numbers):
    sorted_nums = sorted(numbers)
    n = len(sorted_nums)
    if n % 2 == 0:
        return (sorted_nums[n//2 - 1] + sorted_nums[n//2]) / 2
    else:
        return sorted_nums[n//2]

def calc_mode(numbers):
    freq = {}
    for num in numbers:
        freq[num] = freq.get(num, 0) + 1
    max_freq = max(freq.values())
    return [k for k, v in freq.items() if v == max_freq]

def calc_variance(numbers):
    mean = calc_mean(numbers)
    return sum((x - mean)**2 for x in numbers) / (len(numbers) - 1)

def calc_std(numbers):
    return calc_variance(numbers) ** 0.5

def calc_range(numbers):
    return calc_max(numbers) - calc_min(numbers)

age_data = df['Age'].dropna().astype(int).tolist()

results = {
    'Mean': calc_mean(age_data),
    'Minimum': calc_min(age_data),
    'Maximum': calc_max(age_data),
    'Median': calc_median(age_data),
    'Mode': calc_mode(age_data),
    'Variance': calc_variance(age_data),
    'Standard Deviation': calc_std(age_data),
    'Range': calc_range(age_data)
}
results_df = pd.DataFrame({
    'Statistic': list(results.keys()),
    'Value': list(results.values())
})

results_df
```

Fill Missing Values

```python
df_fill = df.copy()
num_cols = ['Age', 'Scholarship', 'Hipertension', 'Diabetes',
            'Alcoholism', 'Handcap', 'SMS_received']

for col in num_cols:
    df_fill[col] = df_fill[col].fillna(df_fill[col].mean())

cat_cols = ['Gender', 'Neighbourhood', 'No-show']

for col in cat_cols:
    df_fill[col] = df_fill[col].fillna(df_fill[col].mode()[0])
```

Remove Duplicate Data

```python
df.drop_duplicates(inplace=True)
```

Handle Incorrect Data Types

```python
df['PatientId'] = df['PatientId'].astype(str)

df['ScheduledDay'] = pd.to_datetime(df['ScheduledDay'])
df['AppointmentDay'] = pd.to_datetime(df['AppointmentDay'])
```

Handle Outliers


```python
Q1 = df['Age'].quantile(0.25)
Q3 = df['Age'].quantile(0.75)
IQR = Q3 - Q1

df = df[(df['Age'] >= Q1 - 1.5 * IQR) &
        (df['Age'] <= Q3 + 1.5 * IQR)]
```

Standardize / Normalize Data

```python
df['Age'] = (df['Age'] - df['Age'].mean()) / df['Age'].std()
```

```python
emp = pd.DataFrame({
    'EmpId': [101, 102, 103],
    'Name': ['Raj','Sham','Prabhakar']
})

rating = pd.DataFrame({
    'EmpId': [101, 102, 103],
    'Rating': ['Excellent','Good','Very Good']
})

emp_data = pd.concat([emp, rating['Rating']], axis=1)

print(emp_data)
```

Key-based Data Integration (Joins)

```python
students = pd.DataFrame({
    'StudentID': [1, 2, 3],
    'Name': ['Amit', 'Neha', 'Rahul']
})

results = pd.DataFrame({
    'StudentID': [1, 2],
    'CGPA': [8.5, 9.1]
})

data_inner = pd.merge(students, results, on='StudentID', how='inner')

print(data_inner)
```

Types of Joins

```python
pd.merge(students, results, on='StudentID', how='inner')   # Common records
pd.merge(students, results, on='StudentID', how='left')    # All left + matches
pd.merge(students, results, on='StudentID', how='right')   # All right + matches
pd.merge(students, results, on='StudentID', how='outer')   # All records
```

Integration Using Index Using join()

```python
df1 = students.set_index('StudentID')
df2 = results.set_index('StudentID')

integrated_data = df1.join(df2)

print(integrated_data)
```
